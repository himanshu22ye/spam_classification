# SPAM-HAM CLASSIFICATION

### SPAM CLASSIFICATION USING BOW,TF-IDF

```python
import pandas as pd
```

```python
import re
import nltk
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
port=PorterStemmer()
```

```python
messages = pd.read_csv(
    'data/SMSSpamCollection.txt',
    sep='\t',
    names=['Label', 'Message']
)

messages.head()
```

```python
corpus=[]
for i in range(0,len(messages)):
    review=re.sub('[^a-zA-Z]',' ',messages['Message'][i])
    review=review.lower()
    review=review.split()
    review=[port.stem(word)for word in review if not word in stopwords.words('english')]
    review=' '.join(review)
    corpus.append(review)
```

### Create BAG OF WORD

```python
#Creating bag of word model
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer(max_features=2500,ngram_range=(1,2))
```

```python
#applying bag of word on corpus
x=cv.fit_transform(corpus).toarray()
```

```python
import numpy as np
np.set_printoptions(edgeitems=30,linewidth=100000,formatter=dict(float=lambda x: "%.3g" % x))
x
```

```python
#output features
y=pd.get_dummies(messages['Label']).astype(int)
```

```python
y=y.iloc[:,0].values
```

```python
#Train test split
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25)
```

```python
#Training the model
from sklearn.naive_bayes import MultinomialNB       #naive_baye because it works well on sparse data 
spam_detect_model=MultinomialNB().fit(x_train,y_train)
```

```python
y_pred=spam_detect_model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score,classification_report
```

```python
accuracy_score(y_test,y_pred)
```

```python
print(classification_report(y_test,y_pred))
```
