# SPAM-HAM CLASSIFICATION

### SPAM CLASSIFICATION USING BOW,TF-IDF

```python
import pandas as pd
```

```python
import re
import nltk
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
port=PorterStemmer()
```

```python
messages = pd.read_csv(
    'data/SMSSpamCollection.txt',
    sep='\t',
    names=['Label', 'Message']
)

messages.head()
```

```python
corpus=[]
for i in range(0,len(messages)):
    review=re.sub('[^a-zA-Z]',' ',messages['Message'][i])
    review=review.lower()
    review=review.split()
    review=[port.stem(word)for word in review if not word in stopwords.words('english')]
    review=' '.join(review)
    corpus.append(review)
```

```python
#output features
y=pd.get_dummies(messages['Label']).astype(int)
y=y.iloc[:,0].values
```

```python
#Train test split
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(corpus,y,test_size=0.25)
```

### Create BAG OF WORD

```python
#Creating bag of word model
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer(max_features=2500,ngram_range=(1,2))
```

```python
#applying bag of word on corpus
x_train=cv.fit_transform(x_train).toarray()
x_test=cv.transform(x_test)
```

```python
import numpy as np
np.set_printoptions(edgeitems=30,linewidth=100000,formatter=dict(float=lambda x: "%.3g" % x))
```

```python
#Training the model
from sklearn.naive_bayes import MultinomialNB       #naive_baye because it works well on sparse data 
spam_detect_model=MultinomialNB().fit(x_train,y_train)
```

```python
y_pred=spam_detect_model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score,classification_report
```

```python
accuracy_score(y_test,y_pred)
```

```python
print(classification_report(y_test,y_pred))
```

## SPAM Classification using TF-IDF

### create tf-idf model

```python
from sklearn.model_selection import train_test_split
z_train,z_test,y_train,y_test=train_test_split(corpus,y,test_size=0.20)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tf=TfidfVectorizer(max_features=2500,ngram_range=(1,2))
z_train=tf.fit_transform(z_train).toarray()
z_test=tf.transform(z_test)
```

```python
from sklearn.naive_bayes import MultinomialNB
spam_tfidf_model=MultinomialNB().fit(z_train,y_train)
```

```python
y_predd= spam_tfidf_model.predict(z_test)
```

```python
from sklearn.metrics import accuracy_score,classification_report
```

```python
score=accuracy_score(y_test,y_predd)
```

```python
print(score)
```

```python
print(classification_report(y_test,y_predd))
```
